```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('athlete_data.csv')
fig=plt.figure(1,figsize=(12,6))
ax=fig.add_subplot(projection='3d')
scatter=ax.scatter(df['Age'],df['Training_Hours_Per_Week'],df['Performance_Score'],s=df['Fatigue_Score'],c=df['Fatigue_Score'],cmap='plasma',label='Age ,Performance Score and Training Hours')
ax.set_xlabel('Age')
ax.set_ylabel('Training Hours Per Week ')
ax.set_zlabel('Performance Score')
ax.set_title('Training Hours and Performance Score by Age')
ax.legend(fontsize=6)
ax.grid(alpha=0.4)
fig.colorbar(scatter,ax=ax)
plt.savefig('Scatter.jpg',dpi=300,bbox_inches='tight')
```
