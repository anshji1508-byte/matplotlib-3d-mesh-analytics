```python
df=pd.read_csv('athlete_data.csv')
fig=plt.figure(1,figsize=(12,6))
ax=fig.add_subplot(111)
Pivot=df.pivot_table(values='ACL_Risk_Score',columns='Weight_kg',index='Height_cm',aggfunc='mean').interpolate(method='linear',axis=0,limit_direction='both').interpolate(method='linear',axis=1,limit_direction='both').fillna(df['ACL_Risk_Score'].mean())
x=Pivot.index
y=Pivot.columns

X,Y=np.meshgrid(x,y)
Z=Pivot.values

contour=ax.contourf(X,Y,Z,cmap='viridis')
fig.colorbar(contour,ax=ax)
ax.set_xlabel('Height (cm)')
ax.set_ylabel('Weight (kg)')
ax.set_title('Height and Weight with ACL Risk Score')
ax.clabel(CS=contour,inline=True,fontsize=8,colors='black')
plt.savefig('Contour.jpg',dpi=300,bbox_inches='tight')
```
