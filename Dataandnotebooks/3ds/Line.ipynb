```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('athlete_data.csv')
fig=plt.figure(1,figsize=(12,6))
ax=fig.add_subplot(projection='3d')
df_sorted=df.sort_values(by='Age',ascending=True)
ax.plot(df_sorted['Age'],df_sorted['Load_Balance_Score'],df_sorted['ACL_Risk_Score'],color='blue',alpha=0.5,marker='o',label='Age ,Load Balance and ACL Risk Score')
ax.set_xlabel('Age')
ax.set_ylabel('Load Balance Score')
ax.set_zlabel('ACL Risk Score')
ax.set_title('Load Balance and ACL Risk Scores by Age')
ax.legend(fontsize=8)
plt.savefig('Line.jpg',dpi=300,bbox_inches='tight')
```
