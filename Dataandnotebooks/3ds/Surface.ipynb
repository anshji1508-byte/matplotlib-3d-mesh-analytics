```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('athlete_data.csv')
fig=plt.figure(1,figsize=(12,6))
ax=fig.add_subplot(projection='3d')

x=df['Training_Hours_Per_Week']
y=df['Rest_Between_Events_Days']

X,Y=np.meshgrid(x,y)

Z=0.6*X-(0.9*Y)+4.5
surface=ax.plot_surface(X,Y,Z,cmap='viridis')
fig.colorbar(surface,ax=ax)
ax.set_title('Training Hours ,Rest days and Fatigue Scores')
ax.set_xlabel('Training Hours')
ax.set_ylabel('Rest Between Event Days')
ax.set_zlabel('Mean Fatigue Score')
ax.view_init(elev=30, azim=45)
plt.savefig('Surface.jpg',dpi=300,bbox_inches='tight')
```
