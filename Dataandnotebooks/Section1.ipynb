```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('athlete_data.csv')
df_Worked=df.head(50)

fig,ax=plt.subplots(2,2,figsize=(20,12))
fig.suptitle('Athlete Data Analysis',color='red',fontsize=50,fontweight='bold')
plt.subplots_adjust(left=0.03,right=0.97,top=0.87,bottom=0.13,hspace=0.3,wspace=0.2)

ax[0,0].plot(df_Worked['Athlete_ID'],df_Worked['heartbeat'],marker='o',color='skyblue',label='HeartBeat')
ax[0,0].axhline(y=df_Worked['heartbeat'].mean(),color='red',linestyle='--',label='Average Heartbeat')
index=df_Worked['heartbeat'].idxmax()
x=df_Worked.loc[index,'Athlete_ID']
y=df_Worked.loc[index,'heartbeat']
ax[0,0].annotate(f'Peak Heartbeat',xy=(x,y),xytext=(10,-30),textcoords='offset points',arrowprops=dict(facecolor='black', shrink=0.05, width=4, headwidth=4),fontsize=16)
ax[0,0].set_xlabel('Athlete Id')
ax[0,0].set_ylabel('Heart Beat')
ax[0,0].legend(fontsize=14)
ax[0,0].grid(True,alpha=0.3)
ax[0,0].set_title('Heart Beat Analysis',fontsize=20)


df_worked2=df.groupby('Training_Intensity')['Performance_Score'].mean().reset_index()

ax[0,1].bar(df_worked2['Training_Intensity'],df_worked2['Performance_Score'],color='orange',edgecolor='black',label='Averages of Intensity',alpha=0.7)
ax[0,1].set_ylim((0,7))
ax[0,1].annotate(f"{df_worked2['Performance_Score'][0]:.2f}",xy=(-0.05,df_worked2['Performance_Score'][0]+0.2),fontsize=17)
ax[0,1].annotate(f"{df_worked2['Performance_Score'][1]:.2f}",xy=(0.95,df_worked2['Performance_Score'][1]+0.2),fontsize=17)
ax[0,1].annotate(f"{df_worked2['Performance_Score'][2]:.2f}",xy=(1.95,df_worked2['Performance_Score'][2]+0.2),fontsize=17)
ax[0,1].axhspan(ymin=0,ymax=4,color='red',alpha=0.09,label="Underperforming Zone")
ax[0,1].legend(fontsize=14)
ax[0,1].set_xlabel('Training Intensity')
ax[0,1].set_ylabel('Performance Score')
ax[0,1].set_title('Intensity Wise Performance Score',fontsize=20)

ax[1,0].scatter(df['Height_cm'],df['Weight_kg'],color='purple',alpha=0.6,label="Height and Weight")
ax[1,0].axhline(y=df['Weight_kg'].mean(),color='red',linestyle='-',linewidth=5,label='Average Weight')
ax[1,0].axvline(x=df['Height_cm'].mean(),color='black',linestyle='-',linewidth=5,label='Average Height')
ax[1,0].set_ylim((50,120))
ax[1,0].annotate("Average Metrics Center",xy=(df['Height_cm'].mean(),df['Weight_kg'].mean()),xytext=(10,120),textcoords='offset points',arrowprops=dict(facecolor='black',arrowstyle='->'),fontsize=16)
ax[1,0].set_xlabel('Height (cm)')
ax[1,0].set_ylabel('Weight (kg)')
ax[1,0].legend(fontsize=14)
ax[1,0].set_title('Height and Weight Relationship',fontsize=20)

ax[1,1].hist(df['Age'],bins=15,color='skyblue',edgecolor='black',label='Age')
ax[1,1].axvspan(xmin=20,xmax=30,color='grey',alpha=0.2)
ax[1,1].annotate('Peak Athletic Group',xy=(22,53),fontsize=17)
ax[1,1].legend(fontsize=14)
ax[1,1].set_xlabel('Age Groups')
ax[1,1].set_ylabel('Frequencies')
ax[1,1].set_title('Age Distribution',fontsize=20)
plt.savefig('Section1.jpg',dpi=300,bbox_inches='tight')
```
