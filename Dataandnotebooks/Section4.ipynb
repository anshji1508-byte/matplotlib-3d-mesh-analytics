```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('athlete_data.csv')
```

```python

fig,ax=plt.subplots(2,2,figsize=(20,12))
fig.suptitle('Athlete Data Analysis',color='red',fontsize=50,fontweight='bold')
plt.subplots_adjust(left=0.03,right=0.97,top=0.87,bottom=0.13,hspace=0.3,wspace=0.2)

ax[0,0].scatter(df['Age'],df['heartbeat'],color='purple',alpha=0.5,label='Age and Hearbeat')
ax[0,0].axvline(x=df['Age'].mean(),label=f'Average Age: {df["Age"].mean():.0f}',linestyle='-',color='red',linewidth=6)
ax[0,0].axhline(y=df['heartbeat'].mean(),label=f'Average Heartbeat: {df['heartbeat'].mean():.0f}',linestyle='-',color='green',linewidth=6)
ax[0,0].legend(fontsize=14)
ax[0,0].set_xlabel('Age')
ax[0,0].set_ylabel('Heart Beat')
ax[0,0].set_title('Age and Heartbeat relation',fontsize=20)
ax[0,0].grid(True,alpha=0.3)


df_made=df['Gender'].value_counts()
ax[0,1].bar(df_made.index,df_made.values,edgecolor='black',color='skyblue',label='Gender Count')
ax[0,1].set_ylim((0,350))
ax[0,1].axhline(df_made.values.mean(),color='red',linestyle='--',linewidth=4,label=f'Average Persons: {df_made.values.mean():.1f}')
ax[0,1].axvline(0.5,color='green',linestyle='--',linewidth=4,label='Divider Line')
ax[0,1].set_xlabel('Gender')
ax[0,1].set_ylabel('Count')
ax[0,1].legend(fontsize=14)
ax[0,1].set_title('Gender Wise Count',fontsize=20)
ax[0,1].grid(True,alpha=0.3)


ax[1,0].hist(df['Fatigue_Score'],bins='auto',edgecolor='black',color='orange',label='Fatigue Score')
ax[1,0].set_ylim((0,120))
ax[1,0].axvline(x=df['Fatigue_Score'].mean(),color='skyblue',linestyle='--',linewidth=4,label=f'Average Fatigue Score: {df['Fatigue_Score'].mean():.0f}')
ax[1,0].set_xlabel('Fatigue Score')
ax[1,0].set_ylabel('Frequency')
ax[1,0].set_title('Fatigue Score Distribution',fontsize=20)
ax[1,0].grid(True,alpha=0.3)
ax[1,0].legend(fontsize=14)

ax[1,1].plot(df.head(30)['Performance_Score'],df.head(30)['Team_Contribution_Score'],color='blue',alpha=0.7,label='Performance and Team Contribution')
ax[1,1].set_xlabel('Performance Score')
ax[1,1].set_ylabel('Team Contribution Score')
ax[1,1].axhline(y=df.head(30)['Team_Contribution_Score'].mean(),color='black',linestyle=':',label=f'Average Team Contribution: {df.head(30)['Team_Contribution_Score'].mean():.1f}',linewidth=3)
ax[1,1].axvline(x=df.head(30)['Performance_Score'].mean(),color='red',linestyle=':',label=f'Average Performance Score: {df.head(30)['Performance_Score'].mean():.1f}',linewidth=3)
ax[1,1].legend(fontsize=14)
ax[1,1].set_title('Performance and Team Contribution (in Scores)',fontsize=20)
plt.savefig('Section4.jpg',dpi=300,bbox_inches='tight')
```
